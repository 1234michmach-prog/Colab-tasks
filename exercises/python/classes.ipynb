## Course Class
Define a course class with the followings:
- A `code` attribute.
- A `name` attribute.
- A `credits` attribute.
- A `semester` attribute.
- An `instructor` attribute.
- An `__init__` method.
- A `__str__` method.

```python
class Course:
  def __init__(self, code, name, credits, semester= None, instructor= None):
    self.code= code
    self.name= name
    self.credits= credits
    self.semester= semester
    self.instructor= instructor

  def __str__(self):
    return(f"{self.name} is taking the following: {self.credits} in the {self.semester} semester with {self.instructor}")
```

```python
python2 = Course("B5", "10",15)

data_vis = Course("B100", "Data Visualisation", 10)

print(python2)
```

```python
courses= {python2: 60, data_vis: 75}
gpa= (60*15 + 75*10) / (15+10)

gpa
```
